```python
num= np.array([10,2,4,5,6])
newarry=num.flatten() # will not change the parent array
newarry[1]=23
print(newarry)
print(num)
```

```python
num= np.array([10,2,4,5,6])
newarry=num.ravel() # changing efects parent and child array
newarry[1]=23
print(newarry)
print(num)
```
