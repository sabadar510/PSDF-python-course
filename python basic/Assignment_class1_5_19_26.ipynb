```python
# Product Discount Checker

# Input
product_price = float(input("Enter product price: "))
membership = input("Are you a member? (yes/no): ")

# Rules
if membership == "yes" and product_price > 5000:
    discount = product_price * 0.20
    print("20% discount applied")
    print("Discount Amount:", discount)

elif membership == "no" and product_price > 5000:
    discount = product_price * 0.10
    print("10% discount applied")
    print("Discount Amount:", discount)

else:
    print("No discount")
```

```python
# AI Recruitment Decision System

# Inputs
name = input("Enter candidate name: ")
age = int(input("Enter age: "))
cgpa = float(input("Enter CGPA: "))
skills = int(input("Enter number of skills: "))
experience = int(input("Enter years of experience: "))
expected_salary = int(input("Enter expected salary: "))
communication = int(input("Enter communication score (1-10): "))

print("\n===== Recruitment Result =====")

# Reject Conditions
if age < 18 or cgpa < 2.7 or skills < 3:
    print("Candidate Rejected")

# Fraud Detection
elif expected_salary > 400000 and experience == 0:
    print("Suspicious Candidate Detected")

# Hiring Levels
elif cgpa >= 3.3 and skills >= 7 and communication >= 7:
    print("Selected as Software Engineer")

elif cgpa >= 3.0 and skills >= 5 and experience >= 1:
    print("Selected as Junior Developer")

elif cgpa >= 2.7 and skills >= 3:
    print("Selected as Intern")

else:
    print("No Suitable Position Found")
```

Write a Python program for a Student Information System that performs the following tasks:

Requirements:

1. Take input from the user for a student with the following details:
- Name (string)
- Age (integer)
- Marks (float)
- Is the student enrolled? (boolean: True/False)
2. Use **logical operators (`and`, `or`) to check conditions.

3. Apply the following conditions:
Condition 1 (Pass/Fail using AND):
If marks are **greater than or equal to 50 AND age is greater than or equal to 18**, print:
Student is PASS and eligible for higher studies
Condition 2 (Special Case using OR):
 If marks are **greater than or equal to 80 OR enrolled is True**, print:
Student is Excellent / Active Participant

Condition 3 (Failure Case):
Otherwise print:
Student needs improvement

4. Finally, print all the student details in a **proper formatted output** like this:
----- Student Record -----
Name: Ali
Age: 20
Marks: 75
Enrolled: True
Result: PASS / FAIL / EXCELLENT (based on condition)


```python
# Student Information System

# Input
name = input("Enter student name: ")
age = int(input("Enter age: "))
marks = float(input("Enter marks: "))
enrolled_input = input("Is the student enrolled? (True/False): ")

# Convert string input to boolean
if enrolled_input.lower() == "true":
    enrolled = True
else:
    enrolled = False

# Default result
result = "Needs Improvement"
status = "FAIL"

print("\n----- Evaluation -----")

# Condition 1 (PASS/FAIL using AND)
if marks >= 50 and age >= 18:
    print("Student is PASS and eligible for higher studies")
    status = "PASS"

# Condition 2 (SPECIAL CASE using OR)
if marks >= 80 or enrolled == True:
    print("Student is Excellent / Active Participant")
    result = "EXCELLENT"

# Condition 3 (Failure Case)
if not (marks >= 50 and age >= 18) and not (marks >= 80 or enrolled == True):
    print("Student needs improvement")
    result = "FAIL"

# Final Output
print("\n----- Student Record -----")
print("Name:", name)
print("Age:", age)
print("Marks:", marks)
print("Enrolled:", enrolled)
print("Result:", result)
```
