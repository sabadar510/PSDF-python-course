1. Login Management System
Question
Create a complete Login Management System in Python.
The system should allow users to:
1.	Register a new account
2.	Login into the system
3.	View all registered users
4.	Exit the program


```python
users = {}

# ---------- Register ----------
def register():
    username = input("Enter username: ")

    if username in users:
        print("User already exists!")
    else:
        password = input("Enter password: ")
        email = input("Enter email: ")

        users[username] = {
            "password": password,
            "email": email
        }

        print("Registration successful!")

# Login
def login():
    username = input("Enter username: ")
    password = input("Enter password: ")

    if username in users and users[username]["password"] == password:
        print("Login successful!")
    else:
        print("Invalid username or password")

#View Users
def view_users():
    if len(users) == 0:
        print("No users found")
    else:
        for username in users:
            print("Username:", username, "| Email:", users[username]["email"])

# ---------- Main Menu ----------
while True:
    print("\n--- LOGIN SYSTEM ---")
    print("1. Register")
    print("2. Login")
    print("3. View Users")
    print("4. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        register()

    elif choice == "2":
        login()

    elif choice == "3":
        view_users()

    elif choice == "4":
        print("Goodbye!")
        break

    else:
        print("Invalid choice!")
```

3.	Create a function that accepts a string and returns the number of uppercase letters, lowercase letters, and digits present in it.

```python
def count_characters(text):
    upper = 0
    lower = 0
    digits = 0

    for ch in text:
        if ch.isupper():
            upper += 1
        elif ch.islower():
            lower += 1
        elif ch.isdigit():
            digits += 1

    return upper, lower, digits


# Example
text = input("Enter a string: ")
u, l, d = count_characters(text)

print("Uppercase Letters:", u)
print("Lowercase Letters:", l)
print("Digits:", d)
```

14.	Write a program that stores product names and prices in a dictionary. Create functions to add products, update prices, and calculate total bill.

```python
# Dictionary to store products and prices
products = {}

# Function to add product
def add_product(name, price):
    products[name] = price
    print(f"{name} added successfully!")

# Function to update product price
def update_price(name, new_price):
    if name in products:
        products[name] = new_price
        print(f"{name} price updated successfully!")
    else:
        print("Product not found!")

# Function to calculate total bill
def total_bill(cart):
    total = 0
    for item in cart:
        if item in products:
            total += products[item]
        else:
            print(f"{item} not found in products!")
    return total


# -------- Main Program --------
while True:
    print("\n===== Product Menu =====")
    print("1. Add Product")
    print("2. Update Price")
    print("3. Calculate Bill")
    print("4. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        name = input("Enter product name: ")
        price = float(input("Enter price: "))
        add_product(name, price)

    elif choice == "2":
        name = input("Enter product name: ")
        price = float(input("Enter new price: "))
        update_price(name, price)

    elif choice == "3":
        cart = input("Enter products (comma separated): ").split(",")
        cart = [item.strip() for item in cart]
        print("Total Bill:", total_bill(cart))

    elif choice == "4":
        print("Exiting program...")
        break

    else:
        print("Invalid choice!")
```

17.	Create a banking system using functions where users can deposit, withdraw, and check balance. Prevent withdrawal if balance is insufficient.

```python
# Initial balance
balance = 0

# Function to deposit money
def deposit(amount):
    global balance
    balance += amount
    print(f"Rs.{amount} deposited successfully.")

# Function to withdraw money
def withdraw(amount):
    global balance

    if amount <= balance:
        balance -= amount
        print(f"Rs.{amount} withdrawn successfully.")
    else:
        print("Insufficient Balance!")

# Function to check balance
def check_balance():
    print(f"Current Balance: Rs.{balance}")

# Main menu
while True:
    print("\n===== Banking System =====")
    print("1. Deposit")
    print("2. Withdraw")
    print("3. Check Balance")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        amount = float(input("Enter amount to deposit: "))
        deposit(amount)

    elif choice == "2":
        amount = float(input("Enter amount to withdraw: "))
        withdraw(amount)

    elif choice == "3":
        check_balance()

    elif choice == "4":
        print("Thank you for using the Banking System!")
        break

    else:
        print("Invalid choice! Please try again.")
```

18.	Write a program that takes a sentence from the user and uses functions to count vowels, consonants, spaces, and digits.

```python
# Function to count vowels
def count_vowels(text):
    vowels = "aeiouAEIOU"
    count = 0

    for ch in text:
        if ch in vowels:
            count += 1

    return count


# Function to count consonants
def count_consonants(text):
    vowels = "aeiouAEIOU"
    count = 0

    for ch in text:
        if ch.isalpha() and ch not in vowels: # isalpha chect krta string ma alphabats hain k nhi true false ma output dayta ha
            count += 1

    return count


# Function to count spaces
def count_spaces(text):
    count = 0

    for ch in text:
        if ch == " ":
            count += 1

    return count


# Function to count digits
def count_digits(text):
    count = 0

    for ch in text:
        if ch.isdigit():
            count += 1

    return count


# Input sentence
sentence = input("Enter a sentence: ")

# Function calls
vowels = count_vowels(sentence)
consonants = count_consonants(sentence)
spaces = count_spaces(sentence)
digits = count_digits(sentence)

# Display results
print("\n===== Analysis Result =====")
print("Sentence:", sentence)
print("Vowels:", vowels)
print("Consonants:", consonants)
print("Spaces:", spaces)
print("Digits:", digits)
```

25.	Create a function that takes a tuple of numbers and returns the sum of prime numbers only.

```python
# Function to check prime number
def is_prime(num):
    if num <= 1:
        return False

    for i in range(2, num):
        if num % i == 0:
            return False

    return True


# Function to sum prime numbers in tuple
def sum_of_primes(tpl):
    total = 0

    for num in tpl:
        if is_prime(num):
            total += num

    return total


# Input tuple from user
numbers = tuple(map(int, input("Enter numbers separated by space: ").split()))

# Function call
result = sum_of_primes(numbers)

print("\nSum of prime numbers:", result)
```

27.	Create a password validation system using conditions and functions. Password must contain:
•	At least 8 characters
•	One uppercase letter
•	One digit


```python
# Function to validate password
def validate_password(password):

    # Condition 1: Length check
    if len(password) < 8:
        return "Password must be at least 8 characters long."

    # Flags
    has_upper = False
    has_digit = False

    # Loop through each character
    for ch in password:
        if ch.isupper():
            has_upper = True
        if ch.isdigit():
            has_digit = True

    # Final checks
    if not has_upper:
        return "Password must contain at least one uppercase letter."

    if not has_digit:
        return "Password must contain at least one digit."

    return "Password is valid!"


# Input from user
pwd = input("Enter password: ")

# Function call
result = validate_password(pwd)

print(result)
```

```python
import os

# ---------------- FILE ----------------
FILE_NAME = "students.txt"

# Load data from file
def load_students():
    students = []
    if os.path.exists(FILE_NAME):
        file = open(FILE_NAME, "r")
        for line in file:
            data = line.strip().split(",")
            if len(data) == 5:
                student = {
                    "id": int(data[0]),
                    "name": data[1],
                    "age": int(data[2]),
                    "course": data[3],
                    "marks": float(data[4])
                }
                students.append(student)
        file.close()
    return students


# Save data to file
def save_students(students):
    file = open(FILE_NAME, "w")
    for s in students:
        file.write(f"{s['id']},{s['name']},{s['age']},{s['course']},{s['marks']}\n")
    file.close()


# ---------------- GRADE FUNCTION ----------------
def get_grade(marks):
    if marks >= 80:
        return "A"
    elif marks >= 70:
        return "B"
    elif marks >= 60:
        return "C"
    else:
        return "Fail"


# ---------------- ADD STUDENT ----------------
def add_student(students):
    try:
        sid = int(input("Enter ID: "))

        # Duplicate check
        for s in students:
            if s["id"] == sid:
                print("Error: Duplicate ID!")
                return

        name = input("Enter Name: ")
        age = int(input("Enter Age: "))
        course = input("Enter Course: ")
        marks = float(input("Enter Marks: "))

        if marks < 0 or marks > 100:
            print("Invalid marks!")
            return

        student = {
            "id": sid,
            "name": name,
            "age": age,
            "course": course,
            "marks": marks
        }

        students.append(student)
        save_students(students)
        print("Student added successfully!")

    except ValueError:
        print("Invalid input!")


# ---------------- VIEW STUDENTS ----------------
def view_students(students):
    if not students:
        print("No records found.")
        return

    # Sort by marks (lambda)
    sorted_list = sorted(students, key=lambda x: x["marks"], reverse=True)

    print("\n----- Student List -----")
    for s in sorted_list:
        print(f"ID: {s['id']}, Name: {s['name']}, Marks: {s['marks']}, Grade: {get_grade(s['marks'])}")


# ---------------- SEARCH ----------------
def search_student(students):
    try:
        sid = int(input("Enter ID to search: "))

        for s in students:
            if s["id"] == sid:
                print("\nStudent Found:")
                print(s)
                return

        print("Student not found!")

    except ValueError:
        print("Invalid ID!")


# ---------------- UPDATE ----------------
def update_student(students):
    try:
        sid = int(input("Enter ID to update: "))

        for s in students:
            if s["id"] == sid:
                new_marks = float(input("Enter new marks: "))

                if new_marks < 0 or new_marks > 100:
                    print("Invalid marks!")
                    return

                s["marks"] = new_marks
                save_students(students)
                print("Updated successfully!")
                return

        print("Student not found!")

    except ValueError:
        print("Invalid input!")


# ---------------- DELETE ----------------
def delete_student(students):
    try:
        sid = int(input("Enter ID to delete: "))

        for s in students:
            if s["id"] == sid:
                students.remove(s)
                save_students(students)
                print("Deleted successfully!")
                return

        print("Student not found!")

    except ValueError:
        print("Invalid input!")


# ---------------- MAIN MENU ----------------
def menu():
    students = load_students()

    while True:
        print("\n===== STUDENT MANAGEMENT SYSTEM =====")
        print("1. Add Student")
        print("2. View Students")
        print("3. Search Student")
        print("4. Update Marks")
        print("5. Delete Student")
        print("6. Exit")

        choice = input("Enter choice: ")

        if choice == "1":
            add_student(students)

        elif choice == "2":
            view_students(students)

        elif choice == "3":
            search_student(students)

        elif choice == "4":
            update_student(students)

        elif choice == "5":
            delete_student(students)

        elif choice == "6":
            print("Goodbye!")
            break

        else:
            print("Invalid choice!")


# Run program
menu()
```
