```python
df_customer_trends=pd.read_csv('D:\\AI Advance\\data scraping\\Airline operation and customer trends\\british-airways_customer_trends.csv')
df_customer_trends
```

```python
df_airline_operation=pd.read_csv('airline_operations_scraped_data.csv')
df_airline_operation
```

```python
df_customer_trends.columns
```

```python
rev_cols_to_drop = ["Review_Title",'Airline', "Review_Date", "Review_Text"]
df_rev_clean = df_customer_trends.drop(columns=rev_cols_to_drop, errors="ignore")
df_rev_clean
```

```python
df_rev_clean.dtypes
```

```python
df_rev_clean.isnull().sum()
```

```python
columns = ['Food_Beverages', 'Inflight_Entertainment']

df_rev_clean[columns] = df_rev_clean[columns].fillna(df_rev_clean[columns].median())
```

```python
df_rev_clean.isnull().sum()
```

```python
df_airline_operation.columns
```

```python
ops_cols_to_drop = ["Flight_ID", "Flight_Number", "Cancellation_Reason"]
df_ops_clean = df_airline_operation.drop(columns=ops_cols_to_drop, errors="ignore")
df_ops_clean
```

```python
df_ops_clean.isnull().sum()
```

```python
# Dono datasets ko side-by-side jorhnay ke liye
df_concat_side = pd.concat([df_ops_clean,df_rev_clean], axis=1)
df_concat_side
```

```python
df_concat_side.isnull().sum()
```

```python
rating_cols = [
    'Overall_Rating',
    'Seat_Comfort',
    'Cabin_Staff_Service',
    'Food_Beverages',
    'Inflight_Entertainment',
    'Value_For_Money'
]

for col in rating_cols:
    df_concat_side[col] = df_concat_side.groupby('Airline')[col].transform(
        lambda x: x.fillna(x.median())
    )
```

```python
df_concat_side.isnull().sum()
```

```python
df_concat_side['Recommended'] = df_concat_side['Recommended'].fillna(df_concat_side['Recommended'].mode()[0])
```

```python
# Convert date/time columns
df["Scheduled_Time"] = pd.to_datetime(df["Scheduled_Time"])
df["Actual_Time"] = pd.to_datetime(df["Actual_Time"], errors="coerce")

# Convert integer columns
int_cols = [
    "Delay_Minutes",
    "Overall_Rating",
    "Seat_Comfort",
    "Cabin_Staff_Service",
    "Food_Beverages",
    "Inflight_Entertainment",
    "Value_For_Money"
]

df[int_cols] = df[int_cols].astype("int64")

# Convert string columns
str_cols = [
    "Airline",
    "Origin_Airport",
    "Destination_Airport",
    "Status",
    "Recommended"
]

df[str_cols] = df[str_cols].astype("string")

# Check datatypes
print(df.dtypes)
```

```python
df_concat_side.dtypes
```

```python
df.isnull().sum()
```

```python
df["Actual_Time"] = df["Actual_Time"].fillna(
    df["Scheduled_Time"] + pd.to_timedelta(df["Delay_Minutes"], unit="m")
)
```

```python
df.isnull().sum()
```

```python
df2 = pd.concat([df_concat_side, df], ignore_index=True)
df2
```

```python
df2["Scheduled_Time"] = pd.to_datetime(df2["Scheduled_Time"])
df2["Actual_Time"] = pd.to_datetime(df2["Actual_Time"], errors="coerce")
```
